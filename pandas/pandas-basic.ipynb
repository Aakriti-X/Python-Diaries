# Working with Series

```python
   # creating empty series
ser = pd.Series()
print(ser)
   #simple array
data = np.array(['g','e','e','k','s'])
ser=pd.Series(data)
print(ser)
```

```python
#a simple list
list = ['g','e','e','k','s']
#creating series from a list
ser = pd.Series(list)
print(ser)
```
